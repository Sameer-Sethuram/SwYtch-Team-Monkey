```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("jolts_cleaned.csv")
df.head()
```

```python
jolts_seasadj_rip = [col for col in df.columns if not "Not Seasonally Adjusted" in col and "Rate in Percent" in col]
jolts_seasadj_lit = [col for col in df.columns if not "Not Seasonally Adjusted" in col and "Level in Thousands" in col]

jolts_nseasadj_rip = [col for col in df.columns if "Not Seasonally Adjusted" in col and "Rate in Percent" in col]
jolts_nseasadj_lit = [col for col in df.columns if "Not Seasonally Adjusted" in col and "Level in Thousands" in col]

jolts_sa_rip = df[jolts_seasadj_rip]
jolts_sa_lit = df[jolts_seasadj_lit]
jolts_nsa_rip = df[jolts_nseasadj_rip]
jolts_nsa_lit = df[jolts_nseasadj_lit]

jolts_sa_rip.shape, jolts_sa_lit.shape, jolts_nsa_rip.shape, jolts_nsa_lit.shape
```

```python
jolts_sa_rip.shape[1] + jolts_sa_lit.shape[1] + jolts_nsa_rip.shape[1] + jolts_nsa_lit.shape[1] == df.shape[1] - 2
```

```python
jolts_sa_rip.to_csv("data/jolts_percents_seasonally_adjusted.csv")
jolts_sa_lit.to_csv("data/jolts_counts_seasonally_adjusted.csv")
jolts_nsa_rip.to_csv("data/jolts_percents_not_seasonally_adjusted.csv")
jolts_nsa_lit.to_csv("data/jolts_counts_not_seasonally_adjusted.csv")
```
